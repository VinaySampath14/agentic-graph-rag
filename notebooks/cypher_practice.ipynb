## Setup — connect to your Neo4j AuraDB instance

```python
from dotenv import load_dotenv
import os
from neo4j import GraphDatabase

load_dotenv()  # reads your .env file

driver = GraphDatabase.driver(
    os.environ["NEO4J_URI"],
    auth=(os.environ["NEO4J_USER"], os.environ["NEO4J_PASSWORD"]),
)
driver.verify_connectivity()
print("Connected.")
```

## Lesson 1 — CREATE and MATCH

`CREATE` makes a new node. `MATCH` finds existing nodes.

We'll create one Paper node and then retrieve it.

```python
# CREATE a single Paper node
with driver.session() as session:
    session.run("""
        CREATE (p:Paper {
            arxiv_id: '1706.03762',
            title: 'Attention Is All You Need',
            year: 2017,
            venue: 'NeurIPS'
        })
    """)
print("Paper created.")
```

```python
# MATCH — retrieve it back
with driver.session() as session:
    result = session.run("""
        MATCH (p:Paper {arxiv_id: '1706.03762'})
        RETURN p.title AS title, p.year AS year
    """)
    for row in result:
        print(row["title"], row["year"])
```
